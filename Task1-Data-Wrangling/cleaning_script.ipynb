```python
import pandas as pd
df = pd.read_csv("car_data.csv")

df.head()
df.info()
df.describe(include="all")
df.isna().sum()

```

```python
df.drop_duplicates(inplace=True)
```

```python
# Numeric → median
num_cols = ['engine_cc','max_power_bhp','mileage_kmpl','km_driven','car_price']
for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)

# Categorical → mode
cat_cols = ['brand','model','fuel_type','transmission','location']
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
df['manufacturing_year'] = df['manufacturing_year'].astype(int)
df['engine_cc'] = df['engine_cc'].astype(int)

```

```python
current_year = 2025

df['car_age'] = current_year - df['manufacturing_year']
df['price_per_cc'] = df['car_price'] / df['engine_cc']
df['price_per_km'] = df['car_price'] / df['km_driven'].replace(0,1)
```

```python
df = df[df['car_price'] <= df['car_price'].quantile(0.99)]
df = df[df['km_driven'] <= df['km_driven'].quantile(0.99)]
```

```python
df.describe()
df.isna().sum()
df.head()
```

```python
df.to_csv("car_data_cleaned.csv", index=False)
print("Cleaned dataset saved!")
```
